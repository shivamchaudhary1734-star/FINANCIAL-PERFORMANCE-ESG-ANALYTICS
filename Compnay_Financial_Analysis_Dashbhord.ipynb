```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn  as sns
```

```python
df=pd.read_csv("company_esg_financial_dataset.csv")
df.head()
```

```python
# 1-Dataset mein missing values aur duplicate records kitne hain?
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["GrowthRate"] =df["GrowthRate"].fillna(True)
df["GrowthRate"] =df["GrowthRate"].fillna("Unknown")
```

```python
df.dtypes
```

```python
# Industry-wise total revenue aur average profit margin kya hai?
 # total revenue
df["Revenue"].sum()
```

```python
# average profit margin
df["ProfitMargin"].mean()
```

```python
IWR =df.groupby("Industry")["Revenue"].sum()
IWR
```

```python
IWPM =df.groupby("Industry")["ProfitMargin"].mean()
IWPM
```

```python
# Year-wise revenue aur growth rate ka trend kya hai?
YWRPM =df.groupby("Year").agg(ywr=("Revenue","sum"),YWG=("ProfitMargin","sum")).reset_index()
YWRPM
```

```python
# Kaunsi companies ki profitability sabse zyada aur sabse kam hai?
CP= df.groupby("CompanyName")["ProfitMargin"].mean()
print("Lowest:",CP.idxmin(),CP.min())
print("Highest:",CP.idxmax(),CP.max())
```

```python
# Financial performance aur ESG scores ke beech kya relationship hai?

cols = ["ESG_Overall", "Revenue", "ProfitMargin", "GrowthRate"]

df[cols].corr()
```

```python
# Carbon emissions, water usage aur energy consumption mein kaunsi companies sabse aage hain?
df.columns
CWEC =df.groupby("CompanyName").agg(cc=("CarbonEmissions","sum"),WU=("WaterUsage","sum"),EC=("EnergyConsumption","sum")).reset_index()
CWEC.sort_values("cc", ascending=False).head(5)
CWEC.sort_values("WU", ascending=False).head(5)
CWEC.sort_values("EC", ascending=False).head(5)
```

```python
# Kaun-si companies high revenue ke saath high ESG performance bhi achieve kar rahi hain?
df.columns
CRE =df.groupby("CompanyName").agg(RS=("Revenue","sum"),ES=("ESG_Overall","mean")).reset_index()
CRE
```

```python
CR= df.groupby("CompanyName")["Revenue"].mean().sort_values().tail(1)
CR
```

```python
# Kaun-si companies high revenue ke saath high ESG performance bhi achieve kar rahi hain?
CRE = df.groupby("CompanyName").agg(
    Total_Revenue=("Revenue", "sum"),
    ESG_Score=("ESG_Overall", "mean")
).reset_index()
CRE.sort_values(
    ["Total_Revenue", "ESG_Score"],
    ascending=False
).head(10)
```
